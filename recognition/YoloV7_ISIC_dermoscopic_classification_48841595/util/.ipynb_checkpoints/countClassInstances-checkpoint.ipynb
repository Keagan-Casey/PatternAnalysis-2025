```python
from pathlib import Path
import csv
srcDirectory = Path("/Users/keagancasey/Desktop/data")
data = {
    "MEL" : 0,
    "NV" : 0,
    "BCC" : 0,
    "AKIEC" : 0,
    "BKL" : 0,
    "DF" : 0,
    "VASC" : 0,
}

classes = list(data.keys())

for folder in srcDirectory.glob("*Anno"):
    for file in folder.glob("*.csv"):
        with open(file, newline='', encoding='utf-8') as csvfile:
            reader = csv.reader(csvfile)
            next(reader)
            for row in reader:
                for cls in data.keys():
                    if float(row[classes.index(cls) + 1]) == 1.0:
                        data[cls] += 1
                        break

print(data)


                    
                    

```
