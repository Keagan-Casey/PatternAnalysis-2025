```python
from pathlib import Path
```

```python
srcDirectory = Path("graffiti/frames/sydney_annotated")
data = {}
classes = []
lineCount = 0
i=0
```

```python
for file in srcDirectory.glob("classes.txt"):
    with open(file, 'r') as file:
        for line in file:
            classes.append(line.strip())
            data[str(lineCount)] = 0
            lineCount +=1
```
