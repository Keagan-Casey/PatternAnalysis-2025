```python
srcDirectory = Path("/Users/keagancasey/Desktop/data")
destinationDir = Path("/Users/keagancasey/Desktop/data/newData")
# Initialise counters for each class
class_counts = {
"MEL": 0,
"NV": 0,
"BCC": 0,
"AKIEC": 0,
"BKL": 0,
"DF": 0,
"VASC": 0
}

class_to_paths = {
    "MEL": MELpaths,
    "NV": NVpaths,
    "BCC": BCCpaths,
    "AKIEC": AKIECpaths,
    "BKL": BKLpaths,
    "DF": DFpaths,
    "VASC": VASCpaths
}


for folder in srcDirectory.glob("*Anno") :
    for file in folder.glob("*.csv"):
        with open(file, newline='', encoding='utf-8') as csvfile:
            reader = csv. reader (csvfile)
            next (reader)
            for row in reader:
                for cls in list(class_counts.keys()):
                    if float(row[list(class_counts.keys()).index(cls) +1]) == 1.0:
                        class_counts [cls] += 1
                        matches = list(srcDirectory.glob(f"*/{row[0]}.jpg"))
                        if matches:
                            class_to_paths[cls].append(matches[0])
                        else:
                            print(f"Image not found for ID {image_id}")
                        break








                        


print(class_counts)


# Arrays to hold full file paths
MELpaths = []
NVpaths = []
BCCpaths = []
AKIECpaths = []
BKLpaths = []
DFpaths = []
VASCpaths = []


folders = {
    "val": destinationDir / "val",
    "test": destinationDir / "test",
    "train": destinationDir / "train"
}

for f in folders.values():
    f.mkdir(parents=True, exist_ok=True)


#print(class_to_paths)


def split_and_copy(class_name, paths):
    random.shuffle(paths)
    total = len(paths)
    if total == 0:
        return

    n_val = max(1, int(0.1 * total))
    n_test = max(1, int(0.1 * total))
    n_train = total - n_val - n_test

    splits = {
        "train": paths[:n_train],
        "val": paths[n_train:n_train + n_val],
        "test": paths[n_train + n_val:]
    }

    for split, imgs in splits.items():
        for img in imgs:
            dst = folders[split] / img.name
            shutil.copy2(img, dst)

    print(f"{class_name}: train={n_train}, val={n_val}, test={n_test}")


for cls, array in class_to_paths.items():
    split_and_copy(cls, array)    
    
```
